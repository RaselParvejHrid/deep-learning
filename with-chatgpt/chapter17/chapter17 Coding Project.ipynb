# Project 17 — Training Stability Laboratory

## Project Description

Build an experiment designed specifically to demonstrate training instability.

Use the same dataset and architecture while changing one factor at a time.

**Experiment A — Learning Rate**

Compare several learning rates.

For example:
- 0.0001
- 0.001
- 0.01
- 0.1

Observe:
- training loss,
- validation loss,
- accuracy,
- gradient norms.

**Experiment B — Input Scaling**

Train once with appropriately scaled features and once without scaling.

Compare convergence.

**Experiment C — Gradient Clipping**

Train a model under conditions where gradients become unusually large.

Compare:
```
loss.backward()
optimizer.step()
```

against:
```
loss.backward()

torch.nn.utils.clip_grad_norm_(
    model.parameters(),
    max_norm=1.0
)

optimizer.step()
```

**Experiment D — Early Stopping**

Implement:

- validation loss monitoring,
- patience,
- best-model checkpointing.

Report:

- The epoch with the best validation loss.
- The epoch at which training actually stopped.
- Final training loss.
- Best validation loss.
- Test performance of the restored best model.

**Engineering Requirement**
Do not simply say:

> "Model A performed better."

Explain why you think it performed differently.

Your analysis should distinguish among:
- Optimization problem
- Data problem
- Numerical stability problem
- Generalization problem
- Architecture problem

This distinction is one of the most important habits you can develop as a future DL engineer.
