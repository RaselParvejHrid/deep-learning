### Helper Functions

```python
def print_tensor_and_attributes(tensor, first_in_cell = False):
    if not first_in_cell:
        print("\n\n")
    print(f"Tensor:\n {tensor}\n")
    print(f"Number of Dimensions: {tensor.ndim}")
    print(f"Shape: {tensor.shape}")
    print(f"Datatype: {tensor.dtype}")
    print(f"Current Device: {tensor.device}")
```

### What is a Tensor?

A tensor is PyTorch's fundamental data structure.

Mathematically:
- Scalar → 0-dimensional tensor
- Vector → 1-dimensional tensor
- Matrix → 2-dimensional tensor
- Higher-dimensional arrays → higher-order tensors

Examples:

$5$ is a scalar.

$[1,2,3]$ is a vector.

$\bigl[ \begin{smallmatrix} 1 & 2 \\ 3& 4 \end{smallmatrix} \bigr]$ is a matrix.

All of these are tensors.

### Tensor Arithmetic

**Vectorized and Element-Wise**

- `a + b`
- `a - b`
- `a * b`

**Matrix Multiplication**
- `a @ b`, or equivalently, `torch.matmul(a, b)`

```python
tensor_a = torch.tensor([[1, 2], [3, 4]])

tensor_b = torch.tensor([[5, 6], [7, 8]])

# Element-Wise Multiplication
print("Element-Wise Product:")
print(tensor_a * tensor_b)

# Matrix Product
print("Matrix Product:")
print(tensor_a @ tensor_b)
```

### Broadcasting

Broadcasting allows tensors of compatible shapes to interact automatically.

```python
x = torch.tensor([[1,2,3], [4,5,6]])

y = torch.tensor([10,20,30])

# x and y are different in shape

# Still the following operation happens as y is broadcast up into the shape of x
z = x + y

print(z)
```

PyTorch conceptually expands y across the rows without physically duplicating the data.

Broadcasting makes code concise and efficient.

### NumPy ↔ PyTorch Interoperability

`torch.from_numpy(numpy_ndarray)` and `tensor.numpy()`.

This interoperability is useful because many scientific Python libraries use NumPy arrays.

```python
import numpy as np

a = [1, 2, 3]
print("Type of 'a':", type(a))

tensor = torch.tensor(a)
print("\nType of 'tensor': ", type(tensor))

to_ndarray = tensor.numpy()
print("\nType of 'to_ndarray':", type(to_ndarray))

back_to_tensor = torch.from_numpy(to_ndarray)
print("\nType of 'back_to_tensor':", type(back_to_tensor))
```
