# Project 2 — Implement an Artificial Neuron


## Problem Statement

**Objective**: Build a single artificial neuron from scratch using NumPy.

Your implementation should:

- Accept an input vector x
- Accept a weight vector w
- Accept a bias b
- Compute $z=w^⊤ x+ b$.

Apply a simple activation function of your choice (for now, you may use the identity function f(z)=z or a step function).

Test the neuron with several different inputs and verify the outputs manually.

The goal is not to build a classifier yet, but to become comfortable with the mechanics of an artificial neuron before we begin training one.

## My Solution

```python
from pprint import pprint

import numpy as np
import pandas as pd
from matplotlib.figure import Figure
from IPython.display import display
import seaborn as sns
```

```python
identity_activation = lambda z : z
```

```python
step_activation = lambda z: 1 if z >= 0 else 0
```

```python
class ArtificialNeuron:
    def __init__(self, w, b, activation):
        self.w = w
        self.b = b
        self.activation = activation

    def activate_on(self, x):
        return self.activation(np.dot(self.w, x) + self.b)
```

```python
w = np.array([1, 2])
b = 3
x = np.array([0, 0])
AN_identity = ArtificialNeuron(w, b, identity_activation)
AN_step = ArtificialNeuron(w, b, step_activation)
pprint(AN_identity.activate_on(x))
pprint(AN_step.activate_on(x))
```

```python
w = np.array([-5, -3, -2])
b = 5
x = np.array([4, -3, 0])
AN_identity = ArtificialNeuron(w, b, identity_activation)
AN_step = ArtificialNeuron(w, b, step_activation)
pprint(AN_identity.activate_on(x))
pprint(AN_step.activate_on(x))
```
