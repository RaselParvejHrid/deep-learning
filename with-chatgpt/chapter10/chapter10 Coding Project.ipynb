## My Solution

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Activation Functions & Derivatives Definition
def step(x):
    return np.where(x >= 0, 1.0, 0.0)

def step_deriv(x):
    # Defined as 0 everywhere (yes, undefined at x=0, but conventionally set to 0)
    return np.zeros_like(x)

def sigmoid(x):
    return 1.0 / (1.0 + np.exp(-x))

def sigmoid_deriv(x):
    s = sigmoid(x)
    return s * (1.0 - s)

def tanh(x):
    return np.tanh(x)

def tanh_deriv(x):
    return 1.0 - np.tanh(x)**2

def relu(x):
    return np.maximum(0.0, x)

def relu_deriv(x):
    # Conventionally set derivative at x=0 to 0
    return np.where(x > 0, 1.0, 0.0)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def leaky_relu_deriv(x, alpha=0.01):
    return np.where(x > 0, 1.0, alpha)

# 2. Evaluation over domain [-5, 5]
x = np.linspace(-5, 5, 1000)

functions = [
    ("Step", step(x), step_deriv(x)),
    ("Sigmoid", sigmoid(x), sigmoid_deriv(x)),
    ("tanh", tanh(x), tanh_deriv(x)),
    ("ReLU", relu(x), relu_deriv(x)),
    ("Leaky ReLU", leaky_relu(x), leaky_relu_deriv(x))
]

# 3. Plotting Functions and Derivatives Side-by-Side
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot Activations
for name, y, _ in functions:
    axes[0].plot(x, y, label=name, linewidth=2)
axes[0].set_title("Activation Functions $f(x)$")
axes[0].set_xlabel("$x$")
axes[0].set_ylabel("$f(x)$")
axes[0].grid(True, alpha=0.3)
axes[0].legend()

# Plot Derivatives
for name, _, dy in functions:
    axes[1].plot(x, dy, label=f"{name} Derivative", linewidth=2)
axes[1].set_title("Derivatives $f'(x)$")
axes[1].set_xlabel("$x$")
axes[1].set_ylabel("$f'(x)$")
axes[1].grid(True, alpha=0.3)
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Report

1. **Saturation**: Saturation occurs when a function's output flatlines for large absolute inputs ($\vert{}x\vert{} \gg 0$), pushing its derivative to zero ($f'(x) \approx 0$).
    - **Sigmoid** and **tanh** suffer from two-sided saturation. When inputs fall outside roughly $[-2, 2]$, the gradient drops near zero, making learning extremely slow.
    - **ReLU** is non-saturating for $x > 0$ (preserving gradient signal), but saturates completely for $x < 0$.
    - **Leaky ReLU** eliminates saturation on both sides by maintaining a constant slope ($\alpha$) for negative inputs.
1. **Vanishing Gradients**: During backpropagation, error signals are calculated using the chain rule, which continuously multiplies local derivatives: $$\frac{\partial \mathcal{L}}{\partial w^{(1)}} = \frac{\partial \mathcal{L}}{\partial a^{(L)}} \prod_{l=1}^{L} \left( W^{(l)T} \cdot f'\left(z^{(l)}\right) \right)$$
   - **Sigmoid** and **tanh**: The maximum value of $f'(x)$ is $0.25$ for Sigmoid and $1.0$ for tanh. In deep networks, repeatedly multiplying by numbers less than $1$ causes the gradient to decay exponentially toward zero, preventing early layers from updating.
   - **ReLU** and **Leaky ReLU**: For positive inputs, the derivative is strictly $1.0$. Gradients pass backward through layers without diminishing in magnitude, allowing successful training of very deep architectures.
1. **Computational Efficiency**
   - **Step**: Extremely fast comparison operation, but unusable for gradient-based training because its derivative is zero everywhere.
   - **Sigmoid** & **tanh**: Computationally heavier due to computing exponential operations ($e^x$).
   - **ReLU** & **Leaky ReLU**: Highly efficient. They require simple conditional checks and scalar operations (np.maximum), making them fast to compute on large tensor inputs.
1. **Typical Use Cases**
   - **Sigmoid**: Output layers for binary classification or multi-label classification (maps logits to probabilities in $[0, 1]$).
   - **tanh**: Recurrent networks (RNNs/LSTMs) and shallow networks where zero-centered output ($(-1, 1)$ range) improves optimization stability.
   - **ReLU**: Standard default activation function for hidden layers in Deep Feedforward Networks and CNNs.
   - **Leaky ReLU** / **Parametric ReLU (PReLU)**: Deep architectures and Generative Adversarial Networks (GANs) where avoiding "dying neurons" (permanently deactivated ReLU units) is critical.
