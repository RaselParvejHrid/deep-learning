### Q12. Draw a rough decision boundary that separates AND gate outputs.

```python
dataset_AND = pd.DataFrame({
    'x1': [0, 0, 1, 1],
    'x2': [0, 1, 0, 1],
    'target' : pd.Categorical([0, 0, 0, 1])
})
```

```python
fig = Figure(figsize=(8, 5))

ax = fig.add_subplot(1, 1, 1)
ax.set_title(f"Feature Space and Decision Boundary for 2-Feature AND Dataset")

# Feature Space
sns.scatterplot(
    data=dataset_AND,
    x='x1',
    y='x2',
    hue='target',
    ax=ax
)

ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')

# Decision Boundary
intercept = 1.5
sns.lineplot(x=[0, intercept], y=[intercept, 0], ax=ax) # Just a convenient Construction

fig.canvas.draw()
display(fig)
```

### Q23. Why can’t a single straight line separate XOR data?

```python
dataset_XOR = pd.DataFrame({
    'x1': [0, 0, 1, 1],
    'x2': [0, 1, 0, 1],
    'target' : pd.Categorical([0, 1, 1, 0])
})
```

```python
fig = Figure(figsize=(8, 5))

ax = fig.add_subplot(1, 1, 1)
ax.set_title(f"Raselian LS on XOR Dataset's Feature Space")

# Feature Space
sns.scatterplot(
    data=dataset_XOR,
    x='x1',
    y='x2',
    hue='target',
    ax=ax
)

ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')


# Raselian LS Lines
sns.lineplot(x=[0, 1], y=[0, 1], ax=ax)
sns.lineplot(x=[0, 1], y=[1, 0], ax=ax)

fig.canvas.draw()
display(fig)
```
