## Problem 2

You are given a two-layer neural network with the following parameters.

Input $X$ is a column vector of size $(2 × 1)$.

$W^{(1)}$ is of shape $(2 × 2)$.  $b^{(1)}$ is of shape $(2 × 1)$.

$W^{(2)}$ is of shape $(2 × 1)$.  $b^{(2)}$ is of shape $(1 × 1)$. 

a) Write the mathematical formula for $Z^{(1)}$.

$$
Z^{(1)} = W^{(1)} X + b^{(1)}
$$

b) Apply ReLU activation to compute $A^{(1)}$.

$$
A^{(1)} = \text{ReLU} \left( Z^{(1)} \right)
$$

c) Write the formula for the final output $\hat{Y}$.

$$
\begin{aligned}
    \hat{Y} &= A^{(2)} \\
    &= \text{ReLU} \left( Z^{(2)} \right) \\
    &= W^{(2)} A^{(1)} + b^{(2)}
\end{aligned}
$$

d) If all weights are initialized to $0.1$ and biases are zero, compute the final output manually for input $X = \begin{bmatrix} 1.0 \\ 2.0 \end{bmatrix}$.

I will assume **ReLU Activation** for hidden layer and **Sigmoid Activation** for Output Layer.

$$
\begin{aligned}
    Z^{(1)}
    &= \begin{bmatrix}
        0.1 &0.1 \\
        0.1 &0.1
    \end{bmatrix}
    \begin{bmatrix}
        1.0 \\ 2.0
    \end{bmatrix}
    +
    \begin{bmatrix}
        0 \\ 0
    \end{bmatrix}
    \\
    &= \begin{bmatrix}
        0.3 \\ 0.3
    \end{bmatrix}
    \\ \\
    A^{(1)}
    &= \text{ReLU}
    \left(
        \begin{bmatrix}
            0.3 \\ 0.3
        \end{bmatrix}
    \right)
    \\
    &= \begin{bmatrix}
            0.3 \\ 0.3
    \end{bmatrix}
    \\ \\
    Z^{(2)}
    &= \begin{bmatrix}
        0.1 &0.1
    \end{bmatrix}
    \begin{bmatrix}
        0.3 \\ 0.3
    \end{bmatrix}
    +
    \begin{bmatrix}
        0
    \end{bmatrix}
    \\
    &= \begin{bmatrix}
        0.06
    \end{bmatrix}
    \\ \\
    \hat{Y} = A^{(2)}
    &= \sigma
    \left( 
        \begin{bmatrix}
            0.06
        \end{bmatrix}
    \right)
    \\
    &= \begin{bmatrix}
            0.515
    \end{bmatrix}
\end{aligned}
$$

## Problem 3

Given $y = 5.0$ and $\hat{y}=3.0$, compute the squared error manually.

$$
\begin{aligned}
    \text{Squared Error}
    &= \left( y - \hat{y} \right)^2
    \\
    &= \left( 5.0 - 3.0 \right)^2
    \\
    &= 4.00
\end{aligned}
$$
